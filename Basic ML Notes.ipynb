# Machine Learning

**Machine Learning (ML) is a branch of Artificial Intelligence (AI) where computers learn patterns from data and use those patterns to make predictions or decisions without being explicitly programmed for every situation.**

## How Machine Learning works

Data → Training → Model → Prediction

 * Collect data
 * Clean and prepare the data
 * Train an ML algorithm
 * Model learns patterns
 * Test the model
 * Use it to make predictions

# Types of Machine Learning

**Machine Learning is mainly divided into 3 major types, with a fourth category often discussed separately**

# Logistic Regression

**Logistic Regression is a Supervised Machine Learning algorithm used for Classification.
Despite the word "Regression", it is mainly used to predict classes/categories.**

## Logistic Regression Equation

**The Logistic Regression equation has two main parts.**

### Linear equation

**z = b0​+b1​X1​+b2​X2​+⋯+bn​Xn​**

* z = linear combination
* b0 = intercept
* b1,b2....,bn = coefficients
* X1,X2.....,Xn =input features

### Sigmoid equation

**The Sigmoid function is a mathematical function used in Logistic Regression to convert any number into a value between 0 and 1.**

**Then convert z into a probability:
P(Y=1)=1/1+e−z**

* σ(z) → Sigmoid function
* z → output of the linear equation
* e → Euler's number, approximately 2.718
* −z → negative of z
* 1 + e⁻ᶻ → denominator

**Suppose the model calculates:
                           z=2**

**Then:
    σ(2)=1/1+e−2**

**Since e
−2
≈0.1353:**

**σ(2)=1/1+0.13531**

**σ(2)≈0.881**

**So the model gives a probability of approximately 88.1%.**

### Important Values to Remember

|  z | Sigmoid | Interpretation      |
| -: | ------: | ------------------- |
| -5 |   0.007 | Very likely Class 0 |
| -2 |   0.119 | Likely Class 0      |
|  0 |   0.500 | Decision boundary   |
|  2 |   0.881 | Likely Class 1      |
|  5 |   0.993 | Very likely Class 1 |

# Dimensionality Reduction 

**Dimensionality Reduction is an Unsupervised Learning technique used to reduce the number of features (variables) in a dataset while preserving as much important information as possible.**

**Imagine you have a dataset with 100 features:**
* Age
* Income
* Credit Score
* Loan Amount
* Spending
...
**100th feature**

* Working with 100 features can make the dataset complex.**
* Dimensionality reduction might reduce:**
* 100 features → 10 important components**
* while retaining much of the useful information.**

**Why do we use it?**

1. Reduce complexity
Fewer features make the dataset easier to work with.

2. Improve visualization
You can reduce 10+ dimensions to 2 dimensions and plot the data.

3. Reduce computation
Fewer features can make ML algorithms faster.

4. Remove redundant information
Highly correlated or less useful information can sometimes be compressed.

5. Help with the Curse of Dimensionality
Very high-dimensional data can make some ML methods less effective.

* PCA — Principal Component Analysis 
**The most important technique to learn first.
Example:
10 original features
↓
PCA
↓
2 Principal Components
Instead of working directly with the original variables, PCA creates new variables called Principal Components that capture as much variance/information as possible.**
* t-SNE
**Mainly used for visualizing high-dimensional data in 2D or 3D.
Commonly used for:
Image data
Text embeddings
Cluster visualization**
* UMAP
**Another technique for reducing high-dimensional data, especially useful for visualization and preserving local structure.**

# Semi Supervised Learning (labeled + unlabeled data)

**Semi-supervised data is a mixture of datasets containing a small amount of labeled data and a very large amount of unlabeled data . It bridges the gap between supervised learning (which requires expensive, manually tagged data) and unsupervised learning (which uses completely raw data)**

 * **The Anatomy of Semi-Supervised Data**

**to understand how this data works, look at its two distinct parts:**
* **The Labeled Core: A tiny fraction (e.g., 5% to 10%) of the dataset. Humans have manually reviewed, verified, and tagged these points with correct answers.**
* **The Unlabeled Pool: The vast majority (e.g., 90% to 95%) of the dataset. This data is cheap and easy to gather in bulk, but it contains no tags, categories, or target answers.**

# 💡 Why Do We Use Semi-Supervised Data?

**Labeling data can be expensive and time-consuming. Semi-supervised learning allows us to use a large amount of easily available unlabeled data along with a smaller labeled dataset.**

**Suppose you have 10,000 loan applications:**

| Data   | Loan Details              | Loan Status |
| ------ | ------------------------- | ----------- |
| 1      | Income, Age, Credit Score | Approved ✅  |
| 2      | Income, Age, Credit Score | Rejected ❌  |
| 3      | Income, Age, Credit Score | Unknown     |
| ...    | ...                       | ...         |
| 10,000 | Income, Age, Credit Score | Unknown     |


**Maybe only 1,000 applications are labeled, while 9,000 are unlabeled.**

**Semi-supervised learning uses both the 1,000 labeled + 9,000 unlabeled records to improve the model.**

**Real-world examples:**
* Medical image classification
* Spam detection
* Fraud detection
* Loan default prediction
* Image and speech recognition

**Semi-supervised learning = Small labeled dataset + Large unlabeled dataset → Better model.**

# Semi-Supervised Learning Algorithms

**1.Main Semi-Supervised Algorithms**
* **Self-Training
Model is first trained using labeled data.
It predicts labels for unlabeled data.
High-confidence predictions are added as new labeled data.
Example: Classification.**
* **Label Propagation
Uses similarities between data points.
Labels are propagated from labeled points to similar unlabeled points.
Commonly used for classification.**
* **Label Spreading
Similar to Label Propagation.
Spreads labels through a graph of data points.
More flexible because it allows labels to change during learning.**
* **Semi-Supervised SVM (S3VM)
Extension of Support Vector Machine.
Uses labeled and unlabeled data to find a better decision boundary.**
* **Pseudo-Labeling
Model predicts labels for unlabeled data.
Predictions with high confidence become pseudo-labels.
These are then used to retrain the model.**

# Reinforcement Learning (RL)

**Reinforcement Learning is a type of Machine Learning where an agent learns by interacting with an environment and receiving rewards or penalties.**

**Simple Example 🎮**

**Imagine an AI playing a game:**

* **AI makes a good move → +10 Reward**
* **AI makes a bad move → −10 Penalty**
* **AI gradually learns which actions produce better results.**

| Component       | Meaning                               |
| --------------- | ------------------------------------- |
| **Agent**       | The learner/decision maker            |
| **Environment** | The world in which the agent operates |
| **State**       | Current situation                     |
| **Action**      | What the agent does                   |
| **Reward**      | Feedback received after an action     |
| **Policy**      | Strategy used to choose actions       |


**RL Process**

**Important Algorithms**

* **Q-Learning – Learns the best action for each state by updating a Q-value based on rewards. It is model-free and commonly used for decision-making.**
* **SARSA – Learns the value of an action based on the current state, current action, reward, next state, and next action.**
* **Deep Q-Network (DQN) – Uses a neural network to estimate Q-values instead of storing them in a table. Useful for complex environments with large state spaces.**
* **Policy Gradient – Directly learns a policy that determines which action the agent should take, rather than learning Q-values.**
* **Actor-Critic – Combines two models: the Actor chooses actions, while the Critic evaluates how good those actions are.**
* **PPO (Proximal Policy Optimization) – Improves the policy gradually while preventing very large updates, making training more stable and reliable.**

**Real-World Applications**

* **🎮 Game playing**
* **🤖 Robotics**
* **🚗 Autonomous vehicles**
* **💹 Trading systems**
* **📦 Recommendation systems**
* **🚁 Drone navigation**
